## Constructing the Tree
Construct the tree in memory

```python
import uuid

from treelib import Tree

from curricula_schema_sdk.course import Course
from curricula_schema_sdk.unit import Unit

tree = Tree()

# Root
tree_id = uuid.uuid4()
tree.create_node(tag="Tree", identifier=tree_id)

# Course
course_id = uuid.uuid4()
tree.create_node(tag="IM-TK", identifier=course_id, parent=tree_id, data=Course(
    subject="IM-TK",
    grade_level="tk",
    curricula_id=course_id,
    language="en",
    isbn=123456789,
    title="IM-TK"
))

# Unit
unit_id = uuid.uuid4()
tree.create_node(tag="IM-TK-unit1", identifier=unit_id, parent=course_id, data=Unit(
    title="Unit 1",
    unit_number=unit_id
))

tree.show()
```

## Exporting the tree

export to simple tree format json file

```python
from curricula_schema_sdk.exporters.stj_exporter import SimpleTreeJsonExporter

stj = SimpleTreeJsonExporter()
stj.export(tree)
```

export to postgres eav tables

```python
from curricula_schema_sdk.exporters.pg_eav_exporter import PostgresEAVExporter

pgeav = PostgresEAVExporter("postgresql://postgres:postgres@localhost:5432/curricula")
pgeav.export(tree)
```
