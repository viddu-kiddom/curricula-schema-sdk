## Constructing the Tree
Construct the tree in memory. This course has two units.

```python

import uuid

from treelib import Tree

from curricula_schema_sdk.course import Course
from curricula_schema_sdk.unit import Unit

tree = Tree()

# Course
course_id = uuid.uuid4()
tree.create_node(tag="IM-TK", identifier=course_id, data=Course(
    subject="IM-TK",
    grade_level="tk",
    curricula_id=course_id,
    language="en",
    isbn=123456789,
    title="IM-TK"
))

# Unit 1
unit_id = uuid.uuid4()
tree.create_node(tag="IM-TK-unit1", identifier=unit_id, parent=course_id, data=Unit(
    title="Unit 1",
    unit_number=unit_id
))

# Unit 2
unit_id = uuid.uuid4()
tree.create_node(tag="IM-TK-unit2", identifier=unit_id, parent=course_id, data=Unit(
    title="Unit 2",
    unit_number=unit_id
))

tree.show(data_property="title")
```

## Exporting the tree
Export the tree to a GIT repo. The nodes are saved to individual `.json` files.

```python
from curricula_schema_sdk.exporters.git_exporter import GitExporter

exporter = GitExporter("./repo", author_name="Viddu Devigere", author_email="viddu@kiddom.co")
```

```python
sha = exporter.export(tree=tree, commit_message="This is a national standard curricula", branch_name="main")
```

## Create a Spanish variant
Create a Spanish variant of the curriculum from the national standard.

```python
import copy

# Make a copy of the tree
spanish_variant = copy.deepcopy(tree)

course = spanish_variant.get_node(spanish_variant.root)
course.data.title = "IM-TK-Spanish"
course.data.language = "es"

spanish_sha = exporter.export(tree=spanish_variant, commit_message="This is a spanish variant", branch_name="spanish",
                base_commit_oid=sha)
```

## Add a new unit to the spanish variant
Add a new unit to the spanish variant of the curricula.

```python
unit3 = Unit(
    title="Unit 3- Spanish",
    unit_number=uuid.uuid4()
)
spanish_variant.create_node("IM-TK-unit3-spanish", "IM-TK-unit3-spanish", parent=course.identifier, data=unit3)
exporter.export(tree=spanish_variant, commit_message="Adding a new unit", branch_name="spanish", base_commit_oid=spanish_sha)
```

## Make a new CA variant from the national tree


```python
ca_variant = copy.deepcopy(tree)

course = ca_variant.get_node(ca_variant.root)
course.data.title = "IM-TK-CA"

ca_sha = exporter.export(ca_variant, commit_message="This is a ca variant", branch_name="California", base_commit_oid=sha)
```
